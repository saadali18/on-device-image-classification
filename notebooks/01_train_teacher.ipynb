```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
log = pd.read_csv("../logs/teacher_train_log.csv")
```

```python
best = log.loc[log["test_top1"].idxmax()]
print(f"Best Epoch  : {int(best['epoch'])}")
print(f"Test Top-1  : {best['test_top1']:.2f}%")
print(f"Test Top-5  : {best['test_top5']:.2f}%")
print(f"Train Top-1 : {best['train_top1']:.2f}%")
print(f"Train Top-5 : {best['train_top5']:.2f}%")
print(f"Train Loss  : {best['train_loss']:.4f}")
print(f"Test Loss   : {best['test_loss']:.4f}")
```

```python
fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(log["epoch"], log["train_top1"], label="Train Top-1")
ax.plot(log["epoch"], log["test_top1"],  label="Test Top-1")
ax.plot(log["epoch"], log["train_top5"], label="Train Top-5")
ax.plot(log["epoch"], log["test_top5"],  label="Test Top-5")
ax.axvline(60, color="grey", linestyle="--", linewidth=0.8)
ax.axvline(80, color="grey", linestyle="--", linewidth=0.8)
ax.set_xlabel("Epoch")
ax.set_ylabel("Accuracy (%)")
ax.set_title("ResNet-110 Teacher — CIFAR-100")
ax.legend()
ax.grid(True)
plt.tight_layout()
plt.savefig("../figs/teacher_training_curve.png", dpi=300)
plt.show()
```
